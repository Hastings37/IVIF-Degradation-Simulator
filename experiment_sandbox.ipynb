```python
import cv2
import os
import numpy as np
from Infrared.ir_lowcontrast import adjust_contrast,DEGRADATION_LEVELS

for (k,v) in DEGRADATION_LEVELS.items():
    print(f"{k}: {v}")
    

img_path='E:\Recurrence\IVIF-Degradation-Simulator\ir_gray.png'

img=cv2.imread(img_path) # 默认按照bgr的形式读取到其内容； 
for level in DEGRADATION_LEVELS:
    adjusted_img=adjust_contrast(img,level)
    save_path=os.path.join('E:\Recurrence\IVIF-Degradation-Simulator\results',f'ir_contrast_{level}.png')
    cv2.imwrite(save_path,adjusted_img)
```
